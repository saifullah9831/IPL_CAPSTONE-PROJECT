3.TOSS WINNER VS MATCH WINNER

```python
count=df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage=(count*100)/df.shape[0]
percentage.round(2)

```

```python
percentage=percentage.round(2)
if percentage >50:
    print("high chance ")
elif percentage < 50:
    print("low chance")
else:
    print("Equal Chance")

print("winnig chance is",percentage)
```

2.2 Top Scores

```python
high=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)## Calculate total runs scored by each player
high
```

```python
high.plot(kind='bar', color='orange')

plt.xlabel('Player')
plt.ylabel('Total Runs')
plt.title('Top 2 Run Scorers')
plt.show()
```

3.10 best bowlling figures

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind = 'barh')
```
